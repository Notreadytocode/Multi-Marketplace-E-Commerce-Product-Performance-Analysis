```python
all_products['discount_pct'] = np.where(
    (all_products['original_price'] > 0) &
    (all_products['original_price'].notna()) &
    (all_products['price'].notna()),
    (
        (all_products['original_price'] - all_products['price'])
        / all_products['original_price']
    ) * 100,
    np.nan
)
```

```python
query = """
WITH ranked_products AS (

    SELECT
        marketplace,
        product_id,
        product_name,
        brand,
        category,
        price,
        rating,
        review_count,
        number_sold,
        gmv,

        ROW_NUMBER() OVER (
            PARTITION BY marketplace
            ORDER BY
                CASE
                    WHEN number_sold IS NOT NULL
                    THEN number_sold
                    ELSE review_count
                END DESC
        ) AS product_rank

    FROM products
)

SELECT
    marketplace,
    product_rank,
    product_name,
    brand,
    category,
    price,
    rating,
    review_count,
    number_sold,
    gmv

FROM ranked_products

WHERE product_rank <= 10

ORDER BY
    marketplace,
    product_rank;
"""

top_10_marketplace = pd.read_sql_query(query, conn)

top_10_marketplace
```

```python
query = """
WITH ranked_products AS (

    SELECT
        marketplace,
        category,
        product_id,
        product_name,
        brand,
        price,
        rating,
        review_count,

        ROW_NUMBER() OVER (
            PARTITION BY marketplace, category
            ORDER BY review_count DESC
        ) AS category_rank

    FROM products

    WHERE category IS NOT NULL
)

SELECT
    marketplace,
    category,
    category_rank,
    product_name,
    brand,
    price,
    rating,
    review_count

FROM ranked_products

WHERE category_rank <= 5

ORDER BY
    marketplace,
    category,
    category_rank;
"""

top_category_products = pd.read_sql_query(query, conn)

top_category_products.head(30)
```

```python
query = """
SELECT
    marketplace,
    product_name,
    brand,
    category,
    price,
    rating,
    review_count,
    number_sold,
    gmv,

    CASE

        WHEN rating >= 4.5
             AND review_count >= 100
             AND number_sold >= 1000
            THEN 'Excellent'

        WHEN rating >= 4.0
             AND review_count >= 50
            THEN 'Strong'

        WHEN rating >= 3.5
            THEN 'Average'

        ELSE 'Needs Improvement'

    END AS performance_segment

FROM products

WHERE rating IS NOT NULL;
"""

performance = pd.read_sql_query(query, conn)

performance.head(30)
```

```python
query = """
WITH product_segments AS (

    SELECT

        marketplace,

        CASE

            WHEN rating >= 4.5
                 AND review_count >= 100
                 AND number_sold >= 1000
                THEN 'Excellent'

            WHEN rating >= 4.0
                 AND review_count >= 50
                THEN 'Strong'

            WHEN rating >= 3.5
                THEN 'Average'

            ELSE 'Needs Improvement'

        END AS performance_segment

    FROM products

    WHERE rating IS NOT NULL
)

SELECT
    marketplace,
    performance_segment,
    COUNT(*) AS product_count

FROM product_segments

GROUP BY
    marketplace,
    performance_segment

ORDER BY
    marketplace,
    product_count DESC;
"""

segments = pd.read_sql_query(query, conn)

segments
```

```python
query = """
SELECT
    marketplace,
    COUNT(*) AS products_with_discount,
    ROUND(AVG(discount_pct), 2) AS avg_discount,
    ROUND(MAX(discount_pct), 2) AS max_discount

FROM products

WHERE discount_pct IS NOT NULL
  AND discount_pct >= 0

GROUP BY marketplace

ORDER BY avg_discount DESC;
"""

discount_analysis = pd.read_sql_query(query, conn)

discount_analysis
```

```python
query = """
SELECT
    marketplace,
    category,
    COUNT(*) AS product_count,
    ROUND(AVG(rating), 2) AS avg_rating,
    ROUND(AVG(price), 2) AS avg_price

FROM products

WHERE category IS NOT NULL
  AND rating IS NOT NULL

GROUP BY
    marketplace,
    category

HAVING COUNT(*) >= 10

ORDER BY
    avg_rating DESC;
"""

best_categories = pd.read_sql_query(query, conn)

best_categories.head(30)
```

```python
query = """
SELECT
    CASE

        WHEN price < 10 THEN 'Under $10'
        WHEN price < 25 THEN '$10-$25'
        WHEN price < 50 THEN '$25-$50'
        WHEN price < 100 THEN '$50-$100'
        ELSE '$100+'

    END AS price_segment,

    COUNT(*) AS product_count,

    ROUND(AVG(rating), 2) AS avg_rating,

    ROUND(AVG(review_count), 2) AS avg_reviews

FROM products

WHERE price IS NOT NULL
  AND rating IS NOT NULL

GROUP BY price_segment

ORDER BY
    CASE price_segment
        WHEN 'Under $10' THEN 1
        WHEN '$10-$25' THEN 2
        WHEN '$25-$50' THEN 3
        WHEN '$50-$100' THEN 4
        WHEN '$100+' THEN 5
    END;
"""

price_rating = pd.read_sql_query(query, conn)

price_rating
```

```python
query = """
WITH product_metrics AS (

    SELECT
        marketplace,
        product_id,
        product_name,
        brand,
        category,
        rating,
        review_count,
        price,

        RANK() OVER (
            PARTITION BY marketplace
            ORDER BY review_count DESC
        ) AS popularity_rank

    FROM products

    WHERE review_count IS NOT NULL
)

SELECT *

FROM product_metrics

WHERE popularity_rank <= 10

ORDER BY
    marketplace,
    popularity_rank;
"""

ranked_products = pd.read_sql_query(query, conn)

ranked_products
```

```python
query = """
SELECT
    marketplace,
    product_name,
    rating,
    review_count,

    RANK() OVER (
        PARTITION BY marketplace
        ORDER BY rating DESC
    ) AS rating_rank,

    DENSE_RANK() OVER (
        PARTITION BY marketplace
        ORDER BY rating DESC
    ) AS dense_rating_rank

FROM products

WHERE rating IS NOT NULL

LIMIT 50;
"""

rank_comparison = pd.read_sql_query(query, conn)

rank_comparison
```
